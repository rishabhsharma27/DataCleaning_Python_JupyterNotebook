```python
df = pd.read_csv("Olympic-Dataset/bios.csv")

df_bios = pd.DataFrame(df)

print(df_bios.info())

df_bios = df_bios.dropna(subset=['Measurements']) 

df_bios = df_bios[df_bios['Measurements'].str.contains('cm') & df_bios['Measurements'].str.contains('kg')]
                                                                           
print(df_bios.info())
```

```python
df[['Height', 'Weight']] = df_bios['Measurements'].str.split('/', expand = True)

print(df)
```

```python
df[['Height', 'Weight']] = df_bios[['Measurements']].apply(
    lambda row: str(row['Measurements']).split('/')  if '/' in str(row['Measurements']) 
    else (['0', row['Measurements']] if 'cm' not in str(row['Measurements']) else ['0', '0']),
    axis=1,
    result_type='expand'  # Ensures the list splits into multiple columns
)

df.drop(['Measurements'],axis=1,inplace= True)

print(df.head())
```

```python
print(df.loc[1:10,[ 'Height', 'Weight']])
```

```python
df = df.dropna(subset=['Height']) 

print(df['Height'].isna().sum() , '\n')
    
df['Height'] = df['Height'].str.replace('cm', ' ')


print(df['Height'])
```

```python
df = df.dropna(subset=['Weight']) 

print(df['Weight'].isna().sum() , '\n')
    
df['Weight'] = df['Weight'].str.replace('kg', ' ')


print(df)
```
